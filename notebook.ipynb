```python
import os
from PIL import Image
import matplotlib.pyplot as plt

# --------------------------------------------------
# Paths
# --------------------------------------------------

ORIGINAL_DIR = "/home/aryan_s2/Detection_with_Distillation/FLIR_ADAS_v2/images_thermal_train/data"
CACHE_DIR    = "/home/aryan_s2/Detection_with_Distillation/Cycle_gan/pseudo_rgb_cache/data"
# Example filename
filename = "video-2Af3dwvs6YPfwSSf6-frame-000000-imW24bapJsHpTahce.jpg"

# Cached image is stored as .png
cache_filename = os.path.splitext(filename)[0] + ".png"

# Full paths
thermal_path = os.path.join(ORIGINAL_DIR, filename)
pseudo_path  = os.path.join(CACHE_DIR, cache_filename)

# --------------------------------------------------
# Check existence
# --------------------------------------------------

if not os.path.exists(thermal_path):
    raise FileNotFoundError(f"Thermal image not found:\n{thermal_path}")

if not os.path.exists(pseudo_path):
    raise FileNotFoundError(f"Pseudo RGB image not found:\n{pseudo_path}")

# --------------------------------------------------
# Load images
# --------------------------------------------------

thermal_img = Image.open(thermal_path)
pseudo_img  = Image.open(pseudo_path)

# --------------------------------------------------
# Display side by side
# --------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Thermal
axes[0].imshow(thermal_img, cmap='gray')
axes[0].set_title("Original Thermal")
axes[0].axis("off")

# Pseudo RGB
axes[1].imshow(pseudo_img)
axes[1].set_title("Pseudo RGB (CycleGAN)")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```
